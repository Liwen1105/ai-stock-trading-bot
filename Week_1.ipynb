```python
import yfinance as yf
import pandas as pd
from pathlib import Path


output_folder = Path("data/raw")
output_folder.mkdir(parents=True, exist_ok=True)
tickers = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'META', 'NVDA', 'TSLA', 'ORCL', 'IBM', 'CRM']
price_data = yf.download(
   tickers=' '.join(tickers),
   start='2019-01-01',
   end=None,  # Until today
   group_by='ticker',
   auto_adjust=True
)


for ticker in tickers:
   df = price_data[ticker].copy().reset_index()
   csv_path = output_folder / f"{ticker}_price.csv"
   parquet_path = output_folder / f"{ticker}_price.parquet"
   df.to_csv(csv_path, index=False)
   df.to_parquet(parquet_path, index=False)
   print(f"Saved: {csv_path}, {parquet_path}")
```

```python
fundamental_data = []


for ticker in tickers:
   stock = yf.Ticker(ticker)
   info = stock.info
   data = {
       'Ticker': ticker,
       'EPS (ttm)': info.get('trailingEps'),
       'P/E (ttm)': info.get('trailingPE'),
       'Revenue (ttm)': info.get('totalRevenue'),
       'Operating Cash Flow (ttm)': info.get('operatingCashflow')
   }
   fundamental_data.append(data)


df_fundamentals = pd.DataFrame(fundamental_data)
df_fundamentals.to_csv("data/raw/tech_fundamentals.csv", index=False)
df_fundamentals.to_parquet("data/raw/tech_fundamentals.parquet", index=False)
df_fundamentals
```
